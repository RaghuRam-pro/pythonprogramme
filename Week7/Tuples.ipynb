##**Operations On Tuple**


```python
t1=(1,2,3)
t2=(4,5,6)
print(t1+t2) #concatenation
print(t1*3) #repitition
print(2 in t1) #membership
print(5 not in t1)
print(t1[1]) #Indexing
print(t2[-1])
print(t1[::-1]) #Slicing
print(len(t1)) #Find Length
print(max(t1)) #Find maximum
print(min(t1)) #Find Minimum
print(t1==t2)#Comparing
for i in t1: #Traversing
  print(i)
```

##Functions                                                                                                      

```python
t=(10,20,30,40)
print(len(t))
print(max(t)) #Find maximum
print(min(t)) #Find Minimum
print(sum(t)) #Sum of Tuple elements
print(sorted(t))#Arranged in Sorted Order
```
