```python
#Arthmetic Operators
a = 20
b = 6
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

```python
#Comparision Operators
a = 10
b = 20
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

##Logical Operators

```python
#AND Operator
age = 20
print(age >= 18 and age<=60)
```

```python
#OR Operator
a = 10
b = 20
print(a > 15 or b > 15)
```

```python
#NOT Operator
age=18
print(not(age==18))
```

```python
#Assignment Operators
a = 10
a += 5
print(a)
a -= 3
print(a)
a *= 2
print(a)
a /= 4
print(a)
```

```python
#Bitwise Operators
a = 5
b = 3
print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

```python
#Membership Operators
name = "Python Programming"
print("Python" in name)
print("Java" in name)
print("Java" not in name)
```

```python
#Identity Operators
a = 100
b = 100
print(a is b)
print(a == b)
```
