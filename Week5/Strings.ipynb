```python
name= "PYTHON"
print(name[0])
print(name[1])
print(name[-1])
print(name[-2])
```

##Slicing

```python
text = "PYTHON PROGRAMMING"
print(text[0:6])
print(text[:6])
print(text[7:])
print(text[:])
print(text[::-1])
print(text[-5:])
```
