```python
#Class and Object
class Student:
    name = "Raghuram"
    age = 20
s1 = Student()
print(s1.name)
print(s1.age)
```

```python
#Constructor
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
s1 = Student("Ram", 20)
print(s1.name)
print(s1.age)
```

```python
#Methods
class Calculator:
    def add(self, a, b):
        return a + b
    def sub(self, a, b):
        return a - b
c = Calculator()
print(c.add(10, 5))
print(c.sub(10, 5))
```
