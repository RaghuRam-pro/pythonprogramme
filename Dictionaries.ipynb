```python
#Creating a Dictionary
student = {
    "name": "Raghu",
    "age": 20,
    "branch": "CSE",
    "marks": 85
}

print(student)
print(type(student))
```

```python
student = {
    "name": "Raghu",
    "age": 20,
    "marks": 85
}

print(student["name"])
print(student["age"])
print(student["marks"])
```

```python
#GET Method()
student = {
    "name": "Raghu",
    "age": 20
}

print(student.get("name"))
print(student.get("age"))
print(student.get("marks"))
print(student.get("marks", 0))
```
